```python
#labelling the data
x=df.drop(['House_Price'],axis=1)
print(x)
y=df['House_Price']
y
```

```python
#splitting te x and y values as train and test data-using train_test_split()
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#linear regression
from sklearn.linear_model import LinearRegression

#linear regression is a class ,o use class first we have to create a object

linear_regressor=LinearRegression()
#watever object name we gave it is a model(Linear_Regression)

#trainig model
linear_regressor.fit(x_train,y_train)
```

```python
#testing model
y_pred_lr=linear_regressor.predict(x_test)
y_pred_lr
```

```python
#model evaluation
#checking results based on y_test

print('R^2 score of LR on Training set',linear_regressor.score(x_train,y_train))
print('R^2 score of LR on Test set',linear_regressor.score(x_test,y_test))
      
```

```python
#checking for error (checking wrong prediction)
from sklearn.metrics import mean_absolute_error as mae,mean_squared_error as mse
print('--Linear Regression Model Evaluation--')
print('Mean Absolute Error',mae(y_test,y_pred_lr))
print('Mean squared Error',mse(y_test,y_pred_lr))
print('root mean squared error',np.sqrt(mse(y_test,y_pred_lr)))
                                             
```

```python
#MAE (8174) → On average, your predictions are wrong by about 8,000 units.
#MSE (101,434,424) → Same errors, but squared. Big errors look even bigger here.
#RMSE (10,071) → Like MAE, but a bit higher because it punishes big mistakes more. So your predictions are usually off by about 10,000 units.
#Think of it like this:
#If you’re guessing someone’s salary, and the real salary is ₹50,000, your model might guess ₹40,000 or ₹60,000. That’s an error of about 10,000 — which matches your MAE/RMSE numbers.
#RMSE means Root Mean Squared Error.
#Think of it like this:
#First, you find the difference between the real value and your model’s prediction.
#You square those differences (so negatives don’t cancel out).
#You take the average of all those squared differences (that’s MSE).
#Finally, you take the square root — that gives RMSE.
# RMSE is just a way of saying:
#“On average, my predictions are off by about this much, measured in the same units as the data.”
#In your case, RMSE ≈ 10,071, which means your model’s guesses are usually wrong by about 10,000 units.
#RMSE means Root Mean Squared Error.
#Think of it like this:
#First, you find the difference between the real value and your model’s prediction.
# You square those differences (so negatives don’t cancel out).
# You take the average of all those squared differences (that’s MSE).
# Finally, you take the square root — that gives RMSE.
#So RMSE is just a way of saying:
#“On average, my predictions are off by about this much, measured in the same units as the data.”
#In your case, RMSE ≈ 10,071, which means your model’s guesses are usually wrong by about 10,000 units.
```
